```python
import pandas as pd 
```

```python
# import inspection set 
inspection_set = pd.read_csv("../../data/processed/inspection_set.csv")
```

```python
inspection_set.info()
```

```python
inspection_set['choice'].value_counts()
```

```python
inspection_set['gt'] = inspection_set['choice'].apply(lambda x: 1 if x == 'Flooded road' else 0)
inspection_set['gt'].value_counts()
```

```python
# add fp and fn column 
# fps are gt = 0 and sentiment_1 = 1, fns are gt = 1 and sentiment_1 = 0 
inspection_set["fp"] = (inspection_set['sentiment_1'] == 1) & (inspection_set['gt'] == 0)
inspection_set["fn"] = (inspection_set['sentiment_1'] == 0) & (inspection_set['gt'] == 1)



```

```python
inspection_set["fp"].value_counts()
```

```python
inspection_set["fn"].value_counts()
```

```python
inspection_set['image'] = inspection_set['image'].str.replace("/data/local-files/?d=/share/XXXX-19/","/share/ju/")
```

```python
inspection_set
```

```python
# copy all fp imgaes to disk 
import shutil
for index, row in inspection_set[inspection_set["fp"] == True].iterrows():
    shutil.copy(row["image"], "../../data/processed/fp_images/")

# copy all fn images to disk 
for index, row in inspection_set[inspection_set["fn"] == True].iterrows():
    shutil.copy(row["image"], "../../data/processed/fn_images/")
```
